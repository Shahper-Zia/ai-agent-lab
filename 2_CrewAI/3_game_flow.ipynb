```python
import random
from crewai.flow.flow import Flow, listen, router, start, and_, or_
from pydantic import BaseModel
```

```python
class GameSession(BaseModel):
    player_won: bool = False
```

```python
class GameFlow(Flow[GameSession]):

    @start()
    def begin_start(self):
        print("Starting the game session")
        player_outcome = random.choice([True, False])
        self.state.player_won = player_outcome
        print( "ID: " ,self.state.id)

    @router(begin_start)
    def check_outcome(self):
        if self.state.player_won:
            return "win"
        else:
            return "lose"

    @listen(check_outcome)
    def log_outcome(self, outcome: str):
        print(f"Player is likely to {outcome}")
    
    @listen("win")
    def celebrate_win(self):
        print("Congratulations!")

    @listen("lose")
    def console_loss(self):
        print("Game Over!")
    
    @listen(or_(celebrate_win, console_loss))
    def end_game(self):
        print("Ending the game session")
        print("Final State:", self.state)
```

```python
flow = GameFlow()
result = flow.kickoff()
```
