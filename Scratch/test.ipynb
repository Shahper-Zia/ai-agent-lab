```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

gemini_api_key = os.getenv("GEMINI_API_KEY")
# print(f"Gemini API Key: {gemini_api_key}")

client = OpenAI(
    api_key=gemini_api_key,  # Google Gemini API key
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"  # Gemini base URL
)

model = "gemini-2.5-flash" # gemini-2.5-flash, gemini-3-pro-preview
# model="gemini-2.5-flash",
```

```python
messages = "what is the capital of france?"

chat_history = [
    {
        "role": "system",
        "content": "You are a helpful assistant."
        "Your task is to answer the user's questions accurately and concisely."
    }
]

chat_history.append({
    "role": "user",
    "content": messages
})

chat_history
```

```python
response = client.chat.completions.create(
    messages=chat_history, 
    model=model,
)
```

```python
print(response.choices[0].message.content)
```

```python
class Demo:
    note = "This is a demo context manager."
    def __init__(self):
        self.agent = []
    def __enter__(self):
        print("ENTER")
        Demo.note = self
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        Demo.note = "Context has been exited."
        print("EXIT")
```

```python
print(Demo.note, type(Demo.note))
with Demo() as demo:
    print(Demo.note.agent, type(Demo.note.agent))
    print("INSIDE")
print(Demo.note, type(Demo.note))
```
