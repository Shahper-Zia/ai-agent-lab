A complete LangGraph-based solution with conditional edges and feedback loops
for analyzing retail competitors in specific locations.

Features:
- Agentic workflow with tool calling
- Conditional routing based on analysis quality
- Feedback loops for iterative refinement
- Conversation persistence
- Real API integration ready

# 1. DEFINE STATE

```python
from typing import Sequence, Annotated, TypedDict
from langchain_core.messages import BaseMessage
from langgraph.graph.message import add_messages

class CompetitorAnalysisState(TypedDict):
    """State for competitor analysis workflow"""
    
    # User inputs
    location: str
    business_type: str
    
    # Gathered data
    competitors_data: dict
    footfall_data: dict
    market_analysis: dict
    
    # Quality tracking
    analysis_confidence: float
    refinement_count: int
    max_refinements: int
    
    # Results
    final_report: str
    execution_time: str
    
    # Conversation history
    messages: Annotated[Sequence[BaseMessage], add_messages]
```

# 3. DEFINE NODES

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings, ChatGoogleGenerativeAI

embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001")
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0.7)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage
from typing import Literal

def gather_data_node(state: CompetitorAnalysisState) -> dict:
    """Node 1: Gather competitor and footfall data"""
    
    print("\n📊 [NODE] Gathering Data...")
    
    competitors_result = search_competitors(state["location"], state["business_type"])
    footfall_result = get_footfall_patterns(state["location"])
    
    return {
        "competitors_data": json.loads(competitors_result),
        "footfall_data": json.loads(footfall_result),
        "messages": [HumanMessage(content="✓ Gathered competitor data and footfall patterns")]
    }


def initial_analysis_node(state: CompetitorAnalysisState) -> dict:
    """Node 2: Perform initial market analysis"""
    
    print("\n🔍 [NODE] Performing Initial Analysis...")
    
    system_prompt = f"""You are an expert retail market analyst. Analyze the following market data:

COMPETITOR DATA:
{json.dumps(state['competitors_data'], indent=2)}

FOOTFALL DATA:
{json.dumps(state['footfall_data'], indent=2)}

Provide a comprehensive market analysis with:
1. Market saturation assessment (0-10 scale)
2. Top 3 competitor strengths
3. Identified market gaps
4. Your confidence level in this analysis (0.0-1.0)

Respond in JSON format with keys: saturation_level, competitor_analysis, gaps, confidence"""
    
    response = llm.invoke([
        SystemMessage(content=system_prompt),
        HumanMessage(content="Analyze this market comprehensively")
    ])
    
    # Parse response
    try:
        result = json.loads(response.content)
        market_analysis = result
        confidence = float(result.get("confidence", 0.5))
    except json.JSONDecodeError:
        market_analysis = {"analysis": response.content}
        confidence = 0.5
    
    return {
        "market_analysis": market_analysis,
        "analysis_confidence": confidence,
        "messages": [HumanMessage(content=f"✓ Initial analysis complete (confidence: {confidence:.0%})")]
    }


def quality_check_node(state: CompetitorAnalysisState) -> Literal["refine_analysis", "generate_report"]:
    """
    CONDITIONAL EDGE: Decide if analysis needs refinement.
    This is a ROUTING FUNCTION that returns which node to visit next.
    
    Returns:
        - "refine_analysis": If confidence is low, refine the analysis
        - "generate_report": If confidence is good, proceed to final report
    """
    
    confidence = state.get("analysis_confidence", 0.5)
    refinements = state.get("refinement_count", 0)
    max_refinements = state.get("max_refinements", 2)
    
    print(f"\n⚖️  [CONDITIONAL EDGE] Quality Check")
    print(f"    Confidence: {confidence:.0%}")
    print(f"    Refinements: {refinements}/{max_refinements}")
    
    # Decision logic
    if confidence < 0.75 and refinements < max_refinements:
        print(f"    Decision: → REFINE (need higher confidence)")
        return "refine_analysis"
    else:
        print(f"    Decision: → GENERATE REPORT")
        return "generate_report"


def refine_analysis_node(state: CompetitorAnalysisState) -> dict:
    """
    Node 3: Refine analysis based on initial insights.
    LOOP: After this node, execution goes back to initial_analysis for deeper insights.
    """
    
    print(f"\n🔄 [NODE] Refining Analysis (Iteration {state['refinement_count'] + 1})...")
    
    system_prompt = f"""You are refining a market analysis. Make it MORE DETAILED and INSIGHTFUL.

Previous Analysis:
{json.dumps(state.get('market_analysis', {}), indent=2)}

Competitor Data:
{json.dumps(state['competitors_data'], indent=2)}

Provide a DEEPER analysis:
1. More specific competitor positioning
2. Detailed market gaps with TAM estimates
3. Concrete, actionable recommendations
4. Higher confidence level (0.0-1.0)

Respond in JSON with keys: refined_analysis, gaps_with_sizing, recommendations, confidence"""
    
    response = llm.invoke([
        SystemMessage(content=system_prompt),
        HumanMessage(content="Refine and deepen this market analysis")
    ])
    
    try:
        result = json.loads(response.content)
        refined_analysis = result
        confidence = float(result.get("confidence", 0.75))
    except json.JSONDecodeError:
        refined_analysis = {"analysis": response.content}
        confidence = 0.75
    
    return {
        "market_analysis": refined_analysis,
        "analysis_confidence": confidence,
        "refinement_count": state.get("refinement_count", 0) + 1,
        "messages": [
            HumanMessage(
                content=f"✓ Refined analysis (iteration {state['refinement_count'] + 1}, confidence: {confidence:.0%})"
            )
        ]
    }


def generate_report_node(state: CompetitorAnalysisState) -> dict:
    """Node 4: Generate final formatted report"""
    
    print("\n📝 [NODE] Generating Final Report...")
    
    # Get opportunities analysis
    competitors_json = json.dumps(state['competitors_data'])
    footfall_json = json.dumps(state['footfall_data'])
    opportunities_json = analyze_market_opportunities(competitors_json, footfall_json)
    opportunities = json.loads(opportunities_json)
    
    # Format the report
    report = f"""
╔══════════════════════════════════════════════════════════════════════════════╗
║                   COMPETITIVE ANALYSIS REPORT                               ║
║          {state['business_type'].title()} Market Analysis - {state['location']}         ║
╚══════════════════════════════════════════════════════════════════════════════╝

📊 EXECUTIVE SUMMARY
────────────────────────────────────────────────────────────────────────────────
• Market Analysis Confidence: {state['analysis_confidence']:.0%}
• Analysis Refinements: {state['refinement_count']}
• Competitors Analyzed: {len(state['competitors_data'].get('competitors', []))}
• Total Competitors in Area: {state['competitors_data'].get('total_competitors_in_area', 'N/A')}

🏢 TOP COMPETITORS
────────────────────────────────────────────────────────────────────────────────
"""
    
    for competitor in state['competitors_data'].get('competitors', [])[:3]:
        report += f"""
{competitor.get('rank')}. {competitor['name']}
   • Location: {competitor['address']}
   • Rating: {competitor['rating']}/5.0 ({competitor['review_count']} reviews)
   • Hours: {competitor['business_hours']}
   • Type: {competitor['store_type']}
   • Contact: {competitor['phone']}
"""
    
    report += """

📍 TRAFFIC & FOOTFALL ANALYSIS
────────────────────────────────────────────────────────────────────────────────
"""
    
    for store in state['footfall_data'].get('stores', [])[:3]:
        report += f"""
{store['store_name']}
   • Avg Daily Visitors: {store['average_daily_visitors']}
   • Peak Day: {store['peak_day']}
   • Peak Hours: {', '.join(store['peak_hours'])}
   • Low Traffic: {', '.join(store['low_traffic_hours'])}
   • Trend: {store['monthly_trend']}
"""
    
    report += f"""

🎯 MARKET INSIGHTS
────────────────────────────────────────────────────────────────────────────────
• Market Saturation: {opportunities.get('market_saturation', {}).get('level', 'N/A')}
• Peak Hours: {state['footfall_data'].get('area_insights', {}).get('busiest_hours_overall', 'N/A')}
• Busiest Day: {state['footfall_data'].get('area_insights', {}).get('busiest_day_overall', 'N/A')}
• Weekend Premium: +{state['footfall_data'].get('area_insights', {}).get('weekend_vs_weekday', '40%')}

💡 IDENTIFIED MARKET GAPS
────────────────────────────────────────────────────────────────────────────────
"""
    
    for i, gap in enumerate(opportunities.get('identified_gaps', []), 1):
        report += f"""
{i}. {gap['gap']}
   • Current Coverage: {gap['current_coverage']}
   • Estimated Demand: {gap['estimated_demand']}
   • Target Market: {gap['target_market']}
   • Investment Potential: {gap['investment_potential']}
"""
    
    report += """

🚀 STRATEGIC RECOMMENDATIONS
────────────────────────────────────────────────────────────────────────────────
"""
    
    for i, strategy in enumerate(opportunities.get('recommended_strategies', []), 1):
        report += f"""
{i}. {strategy['strategy']}
   • Action: {strategy['action']}
   • Expected Impact: {strategy['expected_impact']}
"""
    
    report += f"""

✅ COMPETITIVE ADVANTAGES FOR NEW ENTRANT
────────────────────────────────────────────────────────────────────────────────
"""
    
    for advantage in opportunities.get('competitive_advantages', []):
        report += f"• {advantage}\n"
    
    report += f"""

════════════════════════════════════════════════════════════════════════════════
Report generated with {state['refinement_count']} refinement iteration(s)
Generated at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}
════════════════════════════════════════════════════════════════════════════════
"""
    
    return {
        "final_report": report,
        "messages": [HumanMessage(content="✓ Final report generated successfully")]
    }
```

# 4. BUILD GRAPH WITH CONDITIONAL EDGES & LOOPS

```python
from langgraph.graph import StateGraph, START, END

def build_graph() -> StateGraph:
    """
    Build the complete analysis graph.
    
    Graph Structure:
    ┌─────────────────────────────────────────────────────────────┐
    │ START                                                       │
    │   ↓                                                         │
    │ gather_data                                                 │
    │   ↓                                                         │
    │ initial_analysis                                            │
    │   ↓                                                         │
    │ quality_check  ← [CONDITIONAL EDGE]                         │
    │   ├→ refine_analysis ──→ (LOOP BACK) ──→ initial_analysis   │
    │   └→ generate_report                                        │
    │   ↓                                                         │
    │ END                                                         │
    └─────────────────────────────────────────────────────────────┘
    """
    
    graph = StateGraph(CompetitorAnalysisState)
    
    # Add nodes
    graph.add_node("gather_data", gather_data_node)
    graph.add_node("initial_analysis", initial_analysis_node)
    graph.add_node("refine_analysis", refine_analysis_node)
    graph.add_node("generate_report", generate_report_node)
    
    # Add edges
    graph.add_edge(START, "gather_data")
    graph.add_edge("gather_data", "initial_analysis")
    
    # CONDITIONAL EDGE: Route based on quality check
    graph.add_conditional_edges(
        "initial_analysis",
        quality_check_node,
        {
            "refine_analysis": "refine_analysis",
            "generate_report": "generate_report"
        }
    )
    
    # FEEDBACK LOOP: After refining, go back to analysis
    graph.add_edge("refine_analysis", "initial_analysis")
    
    # Final edge
    graph.add_edge("generate_report", END)
    
    return graph
```

# 5. MAIN EXECUTION

```python
from langgraph.checkpoint.memory import MemorySaver
from IPython.display import display, Image

def run_competitor_analysis(location: str, business_type: str, show_graph: bool = False) -> dict:
    """
    Execute the complete competitor analysis workflow.
    
    Args:
        location: Geographic location (e.g., "Koramangala, Bangalore")
        business_type: Type of business (e.g., "clothing stores")
        show_graph: Whether to display the graph visualization
    
    Returns:
        Dictionary containing the final analysis state
    """
    
    print("\n" + "="*80)
    print("🎯 COMPETITOR ANALYSIS AI PIPELINE")
    print("="*80)
    print(f"\n📍 Location: {location}")
    print(f"🏪 Business Type: {business_type}")
    print(f"⏰ Started: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    
    # Build and compile graph
    graph_builder = build_graph()
    graph = graph_builder.compile(checkpointer=MemorySaver())
    
    if show_graph:
        print("\n" + "-"*80)
        print("\n📈 Workflow Graph Visualization:")
        display(Image(graph.get_graph().draw_mermaid_png()))
        print("-"*80)
    
    # Initialize state
    initial_state = {
        "location": location,
        "business_type": business_type,
        "competitors_data": {},
        "footfall_data": {},
        "market_analysis": {},
        "analysis_confidence": 0.0,
        "refinement_count": 0,
        "max_refinements": 2,
        "final_report": "",
        "execution_time": "",
        "messages": []
    }
    
    print("\n" + "-"*80)
    print("🚀 WORKFLOW EXECUTION")
    print("-"*80)
    
    # Execute graph
    start_time = datetime.now()
    final_state = graph.invoke(
        initial_state,
        config={"configurable": {"thread_id": "competitor_analysis_session_1"}}
    )
    execution_time = (datetime.now() - start_time).total_seconds()
    
    # Display results
    print("\n" + "="*80)
    print("✅ EXECUTION COMPLETE")
    print("="*80)
    print(f"\n📊 Final Confidence: {final_state['analysis_confidence']:.0%}")
    print(f"🔄 Refinement Iterations: {final_state['refinement_count']}")
    print(f"⏱️  Execution Time: {execution_time:.2f} seconds")
    
    print("\n" + final_state["final_report"])
    
    return final_state
```

```python
def interactive_mode():
    """Interactive conversation mode for multiple analyses"""
    
    print("\n" + "="*80)
    print("🤖 COMPETITOR ANALYSIS - INTERACTIVE MODE")
    print("="*80)
    print("\nEnter location and business type to analyze.")
    print("Type 'quit' to exit.\n")
    
    while True:
        location = input("📍 Enter location (e.g., 'Koramangala, Bangalore'): ").strip()
        if location.lower() == 'quit':
            break
        
        business_type = input("🏪 Enter business type (e.g., 'clothing stores'): ").strip()
        if business_type.lower() == 'quit':
            break
        
        if location and business_type:
            run_competitor_analysis(location, business_type)
            print("\n" + "-"*80 + "\n")
        else:
            print("❌ Both location and business type are required.\n")
```
